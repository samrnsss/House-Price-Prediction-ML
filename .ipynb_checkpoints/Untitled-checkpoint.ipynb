```python
import pandas as pd 
import numpy as np 
import seaborn as sns 
import matplotlib.pyplot as plt
```

```python
#Load the data
df = pd.read_csv("HousePricePrediction.csv")
df.shape
df.columns
df.info()
df.describe()
```

```python
#here copy creates independent copies of the selected data so that any change doesnt affect my original data 
#separating datasets 
#selects only the rows where sale price exists
train_df = df[df['SalePrice'].notnull()].copy()  

#selects only the rows where sale price missing
test_df = df[df['SalePrice'].isnull()].copy()   
```

```python
print("Training dataset:", train_df.shape)
print("Testing dataset:", test_df.shape)
```

```python
train_df['SalePrice'].isnull().sum()
```

```python
test_df['SalePrice'].isnull().sum()
```

```python
#data preprocessing of training dataset
train_df.isnull().sum()
train_df.isnull().sum()[train_df.isnull().sum() > 0]

train_df.select_dtypes(include=['int64', 'float64']).columns
```

```python
train_df.select_dtypes(include=['str']).columns
```

```python
print("Numerical columns:", len(train_df.select_dtypes(include=['int64', 'float64']).columns))

print("Categorical columns:", len(train_df.select_dtypes(include=['str']).columns))
```

```python
#remove the id column 
df.drop(['Id'], axis = 1, inplace = True)    # axis = 0 means rows 
```

```python
df
```

```python
train_df = train_df.drop(['Id'], axis = 1)
test_df = test_df.drop(['Id'], axis = 1)
```

```python
print(train_df.shape)
print(test_df.shape)
```

```python
train_df.isnull().sum()
```

```python
train_df.info()
train_df.head()
```

```python
pd.get_dummies()
```

```python
df
```

```python
categorical_cols = train_df.select_dtypes(include = ['str']).columns
print(categorical_cols)
```

```python
#one- hot encoding 
cols = ['MSZoning', 'LotConfig', 'BldgType', 'Exterior1st']
train_df = pd.get_dummies(
    train_df,
    columns = cols,
    drop_first = True,
    dtype = int
)
```

```python
train_df.head()
```

```python
train_df.info()
```

```python
print(train_df.shape)

print(train_df.select_dtypes(include=['object']).columns)
```

```python
#Separating feature(house features) and target(SalePrice)
#X-Independent features  Y-dependent feature (target) = SalePrice

X = train_df.drop(['SalePrice'], axis=1)
y = train_df['SalePrice']      #stores the actual selling prices separately 
```

```python
print("Feature shape:", X)
print("Target shape:", y)
```

```python
#split the dataset
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size = 0.2, random_state = 0
)
```

```python
print("X_train:", X_train.shape)
print("X_test:", X_test.shape)

print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
#training the linear regression model
from sklearn.linear_model import LinearRegression
```

```python
#creating a model
model = LinearRegression()
model.fit(X_train, y_train)
```

```python
print("Model trained successfully")
```

```python
#model evaluation
y_pred = model.predict(X_test)
```

```python
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
from sklearn.metrics import mean_absolute_percentage_error

import numpy as np
```

```python
print("R2 Score:", r2_score(y_test, y_pred))

print("MAE:", mean_absolute_error(y_test, y_pred))

print("RMSE:", np.sqrt(mean_squared_error(y_test, y_pred)))

print("MAPE:", mean_absolute_percentage_error(y_test, y_pred))
```

```python
print("MAPE:", mean_absolute_percentage_error(y_test, y_pred) * 100, "%")
```

```python
#feature scaling

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
```

```python
X_train_scaled = scaler.fit_transform(X_train)

X_test_scaled = scaler.transform(X_test)
```

```python
scaled_model = LinearRegression()

scaled_model.fit(X_train_scaled, y_train)
```

```python
y_pred_scaled = scaled_model.predict(X_test_scaled)
```

```python
print("R2 Score:", r2_score(y_test, y_pred_scaled))

print("MAE:", mean_absolute_error(y_test, y_pred_scaled))

print("RMSE:", np.sqrt(mean_squared_error(y_test, y_pred_scaled)))

print("MAPE:", mean_absolute_percentage_error(y_test, y_pred_scaled) * 100, "%")
```

```python
print("Original Model R2:", r2_score(y_test, y_pred))

print("Scaled Model R2:", r2_score(y_test, y_pred_scaled))
```

```python
print("ORIGINAL MODEL")
print("R2 Score:", r2_score(y_test, y_pred))
print("MAE:", mean_absolute_error(y_test, y_pred))
print("RMSE:", np.sqrt(mean_squared_error(y_test, y_pred)))
print("MAPE:", mean_absolute_percentage_error(y_test, y_pred) * 100, "%")

print("\nSCALED MODEL")
print("R2 Score:", r2_score(y_test, y_pred_scaled))
print("MAE:", mean_absolute_error(y_test, y_pred_scaled))
print("RMSE:", np.sqrt(mean_squared_error(y_test, y_pred_scaled)))
print("MAPE:", mean_absolute_percentage_error(y_test, y_pred_scaled) * 100, "%")
```

# Conclusion

In this assignment, a House Price Prediction model was developed using Supervised Machine Learning and Linear Regression.

The dataset was explored and preprocessed by handling missing values, removing unnecessary columns, and applying One-Hot Encoding to categorical features.

The dataset was divided into training and testing sets, and a Linear Regression model was trained to predict house prices. The model was evaluated using R² Score, Mean Absolute Error (MAE), Root Mean Squared Error (RMSE), and Mean Absolute Percentage Error (MAPE).

The original model achieved an R² score of 0.4365 and a MAPE of 20.65%. After applying feature scaling using StandardScaler, the evaluation results remained almost identical.

This assignment provided practical understanding of data preprocessing, feature encoding, model training, prediction, and evaluation in Supervised Machine Learning.
